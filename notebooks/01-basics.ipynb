# Basic signals: cosine in continuous time and discrete time

Notation (UPM convention):

| Symbol | Meaning | Units |
|---|---|---|
| $\omega$ | angular frequency, continuous time | rad/s |
| $\Omega$ | angular frequency, discrete time | rad/sample |
| $f_s$ | sampling frequency | Hz |
| $T_s = 1/f_s$ | sampling period | s |

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use("dark_background")
```

## 1. Continuous-time cosine

$$
x(t) = \cos(\omega_0 t), \qquad \omega_0 = 2\pi f_0
$$

With $f_0 = 2$ Hz:

$$
\omega_0 = 4\pi \ \text{rad/s}, \qquad T_0 = \frac{1}{f_0} = 0.5 \ \text{s}
$$

Over $T = 2$ s we get $\dfrac{T}{T_0} = 4$ periods.

```python
# Parameters
f0 = 2                  # frequency (Hz)
w0 = 2 * np.pi * f0     # angular frequency (rad/s)
T0 = 1 / f0             # period (s)
T = 2                   # duration (s)

# "Continuous" time: many closely spaced samples
t = np.linspace(0, T, 2000)
x_cont = np.cos(w0 * t)
```

## 2. Discrete-time cosine

We sample $x(t)$ every $T_s$ seconds, i.e. at the instants $t = nT_s$:

$$
x[n] = x(nT_s) = \cos(\omega_0 n T_s) = \cos(\Omega_0 n)
$$

with the discrete-time angular frequency

$$
\Omega_0 = \omega_0 T_s = 2\pi \frac{f_0}{f_s}
$$

With $f_s = 20$ Hz:

$$
T_s = 0.05 \ \text{s}, \qquad \Omega_0 = 2\pi \cdot \frac{2}{20} = 0.2\pi \ \text{rad/sample}
$$

The discrete-time period is

$$
N_0 = \frac{2\pi}{\Omega_0} = \frac{f_s}{f_0} = 10 \ \text{samples}
$$

Since $N_0$ is an integer, $x[n]$ is periodic: it repeats every 10 samples.

```python
# Sampling
fs = 100                 # sampling frequency (Hz)
Ts = 1 / fs             # sampling period (s)
W0 = w0 * Ts            # discrete-time angular frequency (rad/sample)
N0 = 2 * np.pi / W0     # discrete-time period (samples)

n = np.arange(0, int(T * fs) + 1)   # n = 0, 1, ..., 40
x_disc = np.cos(W0 * n)

print(f"Ω0 = {W0 / np.pi:.2f}π rad/sample")
print(f"N0 = {N0:.0f} samples")
```

## 3. Plots

Top: the continuous-time signal, drawn with `plot`.
Bottom: the discrete-time signal, drawn with `stem` — each stem is one sample.

```python
fig, (ax_cont, ax_disc) = plt.subplots(2, 1, figsize=(10, 6))

# Continuous time
ax_cont.plot(t, x_cont)
ax_cont.set_title(r"$x(t) = \cos(\omega_0 t), \quad \omega_0 = 4\pi$ rad/s")
ax_cont.set_xlabel(r"$t$ (s)")
ax_cont.set_ylabel(r"$x(t)$")
ax_cont.grid(alpha=0.3)

# Discrete time
ax_disc.stem(n, x_disc)
ax_disc.set_title(
    rf"$x[n] = \cos(\Omega_0 n), \quad \Omega_0 = 0.2\pi$ rad/sample, "
    rf"$f_s = {fs}$ Hz"
)
ax_disc.set_xlabel(r"$n$")
ax_disc.set_ylabel(r"$x[n]$")
ax_disc.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
