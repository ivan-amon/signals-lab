# Discrete Time Signals

Each one has a different frequency (low, mid and high).

$x_1[n] = \cos\left(\frac{\pi}{20} n\right) \qquad \Omega_0 = \frac{\pi}{20}\ \text{rad/sample}$

$x_2[n] = \cos\left(\frac{\pi}{2} n\right) \qquad \Omega_0 = \frac{\pi}{20}\ \text{rad/sample}$

$x_3[n] = \cos\left(\frac{19\pi}{20} n\right) \qquad \Omega_0 = \frac{\pi}{20}\ \text{rad/sample}$

```python
import matplotlib.pyplot as plt
import numpy as np

n = np.arange(0, 51)

low_omega = np.pi / 20
mid_omega = np.pi / 2
high_omega = 19 * np.pi / 20

x1 = np.cos(low_omega * n)
x2 = np.cos(mid_omega * n)
x3 = np.cos(high_omega * n)

fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(8, 6))

ax1.stem(n, x1, basefmt=' ')
ax1.set_xlabel('n'); ax1.set_ylabel(r"$x_1[n]$"); ax1.grid()

ax2.stem(n, x2, basefmt=' ')
ax2.set_xlabel('n'); ax2.set_ylabel(r"$x_2[n]$"); ax2.grid()

ax3.stem(n, x3, basefmt=' ')
ax3.set_xlabel('n'); ax3.set_ylabel(r"$x_3[n]$"); ax3.grid()


plt.show()
```
