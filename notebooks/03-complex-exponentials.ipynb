## Continuous-Time Complex Exponentials

$x(t) = A\,e^{j \omega_0 t} = A\cos(\omega_0 t) + jA\sin(\omega_0 t)$

Let $\omega_0 = \frac{\pi}{2}\ \text{rad/s}$ and $A = 1$.

```python
import matplotlib.pyplot as plt
import numpy as np

t = np.linspace(0, 10, 2000)
w0 = np.pi / 2
x = 1.5 * np.exp(1j*(w0*t))

fig, axs = plt.subplots(2, 2, figsize=(10, 6), sharex=True)

axs[0, 0].plot(t, x.real)
axs[0, 0].set_ylabel('Re{x(t)}')

axs[0, 1].plot(t, x.imag)
axs[0, 1].set_ylabel('Im{x(t)}')

axs[1, 0].plot(t, np.abs(x))
axs[1, 0].set_ylabel('|x(t)|')

axs[1, 1].plot(t, np.angle(x))
axs[1, 1].set_ylabel('∠x(t) (rad)')

for ax in axs.flat:
    ax.grid()
for ax in axs[1, :]:     
    ax.set_xlabel('t (s)')

fig.tight_layout()
plt.show()
```
